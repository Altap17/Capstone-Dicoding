# Merged df_Log with df_Content

Menggabungkan dataset df_Log dengan df_Content untuk memnuhi kebutuhan AI Engineer

```python
# Merge df_log with df_content to get 'difficulty'
df_merged = pd.merge(
    df_log,
    df_content[["ucid", "difficulty"]],
    on="ucid",
    how="left"
)
df_merged
```

```python
final_features_map = {
    "uuid": "user_id",
    "ucid": "content_topic_id",
    "upid": "problem_question_id",
    "difficulty": "topic_difficulty",
    "total_sec_taken": "time_spent_sec",
    "problem_number": "problem_order",
    "is_correct": "is_correct"
}
df_processed = df_merged[list(final_features_map.keys())].rename(columns=final_features_map)
df_processed
```

Membatasi baris hanya sampai 500 ribu untuk mencegah membesarnya ukuran file data

```python
df_sample_500k = df_processed.sample(n=500000)

# 2. Simpan ke CSV
Processed_Dataset = "Processed_Dataset_500k.csv"
df_sample_500k.to_csv(Processed_Dataset, index=False)
```

Mengubah keterangan dari is_correct menjadi integer

```python
# Convert 'is_correct' boolean to integer (0 or 1)
df_processed["is_correct"] = df_processed["is_correct"].astype(int)
```

Melakukan cleaning data terhadap topic_difficulty dengan mengisinya dengan "unknown"

```python
# Handle missing difficulty values
df_processed["topic_difficulty"] = df_processed["topic_difficulty"].fillna("unknown")
```

```python
# Use df_processed as the base for filtering, which is equivalent to df_log_full
sequence_counts = df_processed.groupby(["user_id", "content_topic_id"])["problem_question_id"].transform("count")
df_log_valid_seq = df_processed[sequence_counts >= 10].copy()

print("Sampling data berdasarkan User utuh...")
# Take up to 10,000 unique users from the valid sequences
sample_users = df_log_valid_seq["user_id"].drop_duplicates().head(10000)

# Filter all learning history from the selected users
df_filtered_users = df_log_valid_seq[df_log_valid_seq["user_id"].isin(sample_users)].copy()

print("Sorting data for LSTM...")
# Sort the data for LSTM to ensure temporal sequence
df_final_dataset = df_filtered_users.sort_values(by=["user_id", "content_topic_id", "problem_order"]).reset_index(drop=True)

print(f"Data siap! Total baris: {len(df_final_dataset)}")
print(df_final_dataset.head())
```

Mengurutukan berdasarkan 10.0000 Unique users

```python
# Take up to 10,000 unique users from the valid sequences
sample_users = df_log_valid_seq["user_id"].drop_duplicates().head(10000)
```

Memfilter berdasarkan data learning history

```python
# Filter all learning history from the selected users
df_filtered_users = df_log_valid_seq[df_log_valid_seq["user_id"].isin(sample_users)].copy()
```

```python
# Sort the data for LSTM to ensure temporal sequence
df_final_dataset = df_filtered_users.sort_values(by=["user_id", "content_topic_id", "problem_order"]).reset_index(drop=True)
```

```python
df_final_dataset
```

```python
df_aggregated = df_final_dataset.groupby(["user_id", "content_topic_id"]).agg(
    total_time_spent=("time_spent_sec", "sum"),
    avg_time_per_problem=("time_spent_sec", "mean"),
    total_problems_attempted=("problem_question_id", "count"),
    total_correct_answers=("is_correct", "sum"),
    success_rate=("is_correct", "mean"),  # Mean of 0/1 gives success rate
    max_problem_order=("problem_order", "max"),  # Max problem number attempted in this topic
    topic_difficulty_label=("topic_difficulty", "first") # Add this line to include the difficulty label
).reset_index()
```

```python
# Ordinal encoding for 'topic_difficulty'
difficulty_mapping = {
    'easy': 1,
    'normal': 2,
    'hard': 3,
    'unknown': 0  # Assign 0 for unknown difficulty
}
```

```python
# Map difficulty from df_content based on content_topic_id
df_aggregated["topic_difficulty_encoded"] = df_aggregated["content_topic_id"].map(
    df_content.set_index("ucid")["difficulty"]
).map(difficulty_mapping).fillna(0).astype(int)
```

```python
if len(df_final_dataset) > 100000:
    df_output_100k = df_final_dataset.sample(n=100000, random_state=42)
    print(f"Extracted 100,000 rows. Final output DataFrame shape: {df_output_100k.shape}")
    print(df_output_100k.head())
else:
    df_output_100k = df_final_dataset
    print(f"Less than 100,000 rows after filtering. Final output DataFrame shape: {df_output_100k.shape}")
    print(df_output_100k.head())
```

# EXPLORATORY DATA ANALYSISI

```python
# histogram: the distribution of grade of the unique students
plt.figure(figsize=(20,5))
g = sns.countplot(x='user_grade', data = df_user)
g.set(xlabel='Grade', ylabel='Frequency',title = 'The distribution of the number of unique students by grade')
plt.show()
```

```python
# Calculate the number of active months for each user
# Convert 'timestamp_TW' to datetime objects in df_log
df_log['timestamp_TW'] = pd.to_datetime(df_log['timestamp_TW'], utc=True)

# Extract year and month from 'timestamp_TW'
df_log['year_month'] = df_log['timestamp_TW'].dt.to_period('M')

# Calculate the number of unique active months for each user
df_user_active_months = df_log.groupby('uuid')['year_month'].nunique().reset_index()
df_user_active_months.rename(columns={'year_month': 'n_active_months'}, inplace=True)

# histogram: the distribution of number of active months of users
plt.figure(figsize=(20,5))
g = sns.countplot(x='n_active_months', data = df_user_active_months)
g.set(xlabel='Number of Active Months', ylabel='Frequency',title = 'The distribution of number of active months of students')
plt.show()
```

# BUSSINESS QUESTION

```python
# Use df_aggregated directly for subsequent analysis
df = df_aggregated
```

```python
plt.figure(figsize=(10, 6))
hard_df = df[df['topic_difficulty_label'] == 'hard'].copy()
# Membagi siswa menjadi dua kelompok berdasarkan median waktu belajar
hard_df['time_group'] = pd.qcut(hard_df['total_time_spent'], q=2, labels=['Waktu Belajar Rendah', 'Waktu Belajar Tinggi'], duplicates='drop')
q1_result = hard_df.groupby('time_group', observed=False)['success_rate'].mean()

sns.barplot(x=q1_result.index, y=q1_result.values, palette='Blues_d', hue=q1_result.index, legend=False)
plt.title('Q1: Rata-rata Tingkat Keberhasilan vs Waktu Belajar (Topik Sulit)', fontsize=14)
plt.ylabel('Rata-rata Success Rate')
plt.xlabel('Kelompok Waktu Belajar')
plt.ylim(0, 1)

for i, val in enumerate(q1_result.values):
    plt.text(i, val + 0.02, f'{val:.2%}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
print(f"Insight Q1: Siswa dengan waktu belajar tinggi pada topik sulit memiliki success rate {q1_result['Waktu Belajar Tinggi']:.2%}, sedikit lrbih kecil dari kelompok rendah.\n")
```

```python
plt.figure(figsize=(10, 6))
# Mendefinisikan kelompok sukses (High Success >= 70%)
df['success_group'] = pd.cut(df['success_rate'], bins=[0, 0.7, 1.0], labels=['Sukses Rendah (<70%)', 'Sukses Tinggi (>=70%)'])
q2_result = df.groupby('success_group')['avg_time_per_problem'].mean()

sns.barplot(x=q2_result.index, y=q2_result.values, palette='Oranges_d')
plt.title('Q2: Rata-rata Waktu per Soal: Sukses Tinggi vs Rendah', fontsize=14)
plt.ylabel('Rata-rata Waktu per Soal (detik)')
plt.xlabel('Kelompok Kesuksesan')

# Menambahkan label nilai pada bar
for i, val in enumerate(q2_result.values):
    plt.text(i, val + 0.5, f'{val:.1f} detik', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
print(f"Insight Q2: Siswa sukses tinggi rata-rata menghabiskan {q2_result['Sukses Tinggi (>=70%)']:.1f} detik per soal, menunjukkan ketelitian yang lebih baik.\n")
```
